# Conducting Bivariate Analysis on Age and Survival

```python
bins = [0, 18, 37, 56, 81]
labels = ['<18', '18-36', '37-55', '>55']
print(f"Bins defined: {bins}")
print(f"Labels defined: {labels}")
```

I decided to class the following **ages by groups** of 18 for my analysis. Classing them as such helped to identify which were children and keep categories consistent.

AI prompt used: Split the age columns into groups of these categories -> 18s, 18-36, 37-55, 57+

```python
df['AgeGroup'] = pd.cut(df['Age'], bins=bins, labels=labels, right=False)
df.head()
```

I then **classed the age of each individual** in the dataset by the classifications I had made.

AI prompt used: apply the defined age groups to the age column in the DataFrame to create a new 'AgeGroup' column.

```python

# Calculate the survival rate for each AgeGroup
survival_rate_by_agegroup = df.groupby('AgeGroup')['Survived'].mean().reset_index()

# Rename the 'Survived' column to 'Survival_Rate' for clarity
survival_rate_by_agegroup = survival_rate_by_agegroup.rename(columns={'Survived': 'Survival_Rate'})

display(survival_rate_by_agegroup)

# Create the bar plot
plt.figure(figsize=(10, 6))
sns.barplot(x='AgeGroup', y='Survival_Rate', data=survival_rate_by_agegroup, palette='viridis')
plt.title('Survival Rate by Age Group')
plt.xlabel('Age Group')
plt.ylabel('Survival Rate')
plt.ylim(0, 1) # Survival rate is between 0 and 1
plt.grid(axis='y', alpha=0.75)
plt.show()
```

I wanted to examine each age group's chance of survival. Hence, I took the **mean** of the surived column for each age group to determine the **rate**.

After visually plotting it on a bar chart, it is evident that those **under 18 had the highest surival chances** on board. Whilst, those **over 55 had the lowest**. This confirms my initial prediction that the older people were the most likely to have died on the Titanic.

However, it is important to note the age column had **missing values** so the results are **not entirely representative** of the whole ship.
